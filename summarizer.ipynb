```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('CNN_Articels_clean.csv')
```

```python
df.head()
```

```python
df.drop(columns= ['Index', 'Author', 'Date published',	'Category', 'Section', 'Url', 'Description', 'Headline', 'Second headline'], inplace=True, axis=1)
```

```python
df['Article text'] = df['Article text'].str.replace('(CNN)', '')
df['Keywords'] = df['Keywords'].str.replace('- CNN', '')
df['Article text'] = df['Article text'].str.strip()
df['Keywords'] = df['Keywords'].str.strip()
```

```python
df.head()
```

```python
df['Keywords'][0]
```

```python
df.drop(columns= ['Keywords'], inplace=True, axis=1)
```

```python
df['Article text'] = df['Article text'].str.lower()
```

```python
from transformers import BartTokenizer, BartForConditionalGeneration
```

```python
model_name = "facebook/bart-large-cnn"

tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name)
```

```python
from tqdm import tqdm
summaries = []

for i in tqdm(df['Article text'][:10]):
    tokenized = tokenizer(i, return_tensors='pt', max_length=1024, truncation=True)
    summary = model.generate(tokenized['input_ids'])
    decoded = tokenizer.decode(summary[0], skip_special_tokens=True)
    summaries.append(decoded)
```
